```python
products = {
    101: {"name": "Laptop", "category": "Electronics", "price": 1200, "stock": 50},
    102: {"name": "Shirt", "category": "Apparel", "price": 25, "stock": 200},
    103: {"name": "Coffee Maker", "category": "Home Appliances", "price": 80, "stock": 30}
}

products[102]["stock"] = 250

print(products)

products[104] = {
    "name": "Smartphone",
    "category": "Electronics",
    "price": 800,
    "stock": 100
}

print(products)
```

```python
age = int(input("Enter your age: "))

if age >= 0 and age <= 12:
    print("Child")

elif age >= 13 and age <= 19:
    print("Teenager")

elif age >= 20 and age <= 64:
    print("Adult")

elif age >= 65:
    print("Senior")

else:
    print("Invalid age")
```

```python
title = input("Enter the title of the book: ")
author = input("Enter the author's name: ")
year = int(input("Enter the year of publication: "))
price = float(input("Enter the price of the book: "))

book = (title, author, year, price)

print("Title:", book[0])
print("Author:", book[1])
print("Year of Publication:", book[2])
print("Price: $", book[3], sep="")
```

```python
num = float(input("Enter a number: "))

if num > 0:
    print("The number is Positive")

elif num < 0:
    print("The number is Negative")

else:
    print("The number is Zero")
```

```python
student_details = {
    "Ali": {"age": 20, "grade": "A", "city": "Karachi"},
    "Sara": {"age": 21, "grade": "B", "city": "Lahore"},
    "Omar": {"age": 19, "grade": "A", "city": "Islamabad"},
    "Hina": {"age": 22, "grade": "C", "city": "Faisalabad"}
}

student_details.update({
    "Zara": {"age": 20, "grade": "B", "city": "Multan"}
})

student_details.pop("Omar")

print(student_details)
```

```python

inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 8},
    "Harry Potter": {"price": 12.50, "quantity": 2}
}

inventory.update({
    "Atomic Habits": {"price": 15.99, "quantity": 4}
})

print("All book titles:", list(inventory.keys()))

for book in list(inventory.keys()):
    if inventory[book]["quantity"] == 0:
        inventory.pop(book)

print("Final inventory:", inventory)
```

```python
letter = input("Enter a single letter: ")

if len(letter) != 1 or not letter.isalpha():
    print("Please enter valid alphabet")

elif letter.lower() in "aeiou":
    print(letter, "is a Vowel")

else:
    print(letter, "is a Consonant")
```
