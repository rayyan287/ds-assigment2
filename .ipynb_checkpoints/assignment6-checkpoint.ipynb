```python
balance = 1000

try:
    amount = float(input("Enter amount to withdraw: "))

    if amount <= 0:
        print("Amount must be more than zero.")

    elif amount > balance:
        print("Insufficient balance.")

    else:
        balance -= amount
        print("Remaining balance:", balance)

except ValueError:
    print("Please enter a valid number.")

finally:
    print("Thank you for using the ATM.")
```

```python
available_seats = 5

try:
    user_seats = int(input("ENTER SEATS YOU WANT"))
    if user_seats <= 0 :
         print("You must book at least 1 ticket.")

    elif user_seats  > available_seats:
         print("Not enough seats available.")

    else:
         available_seats -= user_seats
         print(user_seats, "ticket(s) booked successfully." )

except ValueError:
    print("Please enter a valid number.")

finally:
    print("Thank you for booking with us.")
```

```python
try:
    user_items = int(input("Enter the number of items you want to buy: "))

    if user_items < 1:
        print("You must buy at least one item.")

    else:
        print("You added", user_items, "item(s) to your cart.")

except ValueError:
    print("Please enter a valid quantity.")
```

```python
try:
    day = int(input("Enter a day number (1-7): "))

    if day == 1:
        print("Day is Monday")

    elif day == 2:
        print("Day is Tuesday")

    elif day == 3:
        print("Day is Wednesday")

    elif day == 4:
        print("Day is Thursday")

    elif day == 5:
        print("Day is Friday")

    elif day == 6:
        print("Day is Saturday")

    elif day == 7:
        print("Day is Sunday")

    else:
        print("Please enter a number between 1 and 7.")

except ValueError:
    print("Please enter a valid number.")
```

```python
try:
    days_late = int(input("Enter the number of days late: "))

    if days_late < 0:
        print("Days late cannot be negative.")

    else:
        fine = days_late * 0.50
        print("Total fine: $", format(fine, ".2f"), sep="")

except ValueError:
    print("Please enter a valid number.")
```

```python
class Book:

    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year


book1 = Book("To Kill a Mockingbird", "Harper Lee", 1960)

print(book1.title)
print(book1.author)
print(book1.year)
```

```python
class Book:

    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def display_info(self):
        print(f"'{self.title}' by {self.author}, published in {self.year}.")

    def is_classic(self):
        print("This is a book.")


book1 = Book("To Kill a Mockingbird", "Harper Lee", 1960)

book1.is_classic()
book1.display_info()
```

```python
class Student:

    def __init__(self, name, grade, age):
        self.name = name
        self.grade = grade
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}, I am {self.age} years old.")

    def show_grade(self):
        print(f"My grade is {self.grade}.")


student1 = Student("Alice", "A", 16)

student1.introduce()
student1.show_grade()
```

```python
class Light:

    def turn_on(self):
        print("The light is now ON.")

    def check_status(self):
        print("The light is bright and working.")


light1 = Light()

light1.turn_on()
light1.check_status()
```

```python
available_slots = [101, 102, 103, 104, 105]

reserved = [106, 107, 108, 109, 110]

try:
    slot = int(input("Enter the parking slot number you want to reserve: "))

    if slot in available_slots:
        print(f"Slot {slot} is reserved for you.")

    elif slot in reserved:
        print(f"Slot {slot} is already reserved by someone else.")

    else:
        print(f"Slot {slot} does not exist.")

except ValueError:
    print("Invalid input. Please enter a valid slot number.")
```
