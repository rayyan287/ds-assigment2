```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]

count = 0

for user in users:
    if count == 3:
        print("Limit reached. Stop sending emails.")
        break

    print("Email sent to", user)
    count += 1
```

```python

temps = [22, 25, 30, 45, 28]
critical = 40

for temp in temps:
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break

    print("Temperature:", temp)
```

```python

cart = [100, 250, 50, 300]

total = 0

for price in cart:
    total = total + price

print("Total Price:", total)
```

```python

customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    if customer["paid"] == False:
        continue

    print("Access granted to:", customer["name"])
```

```python

users = ["Ali", "Sara", "John"]

for user in users:
    print("Hello", user + ", your subscription is active!")
```

```python

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

domains = []

for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)

print("domains:", domains)
```

```python

numbers = [1, 3, 5, 9, 10, 12, 15, 17, 20]

new_list = [num for num in numbers if num % 3 == 0 or num % 5 == 0]

print(new_list)
```

```python

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

for num in numbers:
    for multiplier in multipliers:
        result = num * multiplier
        print(num, "x", multiplier, "=", result)
```
