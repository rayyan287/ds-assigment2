```python

class Employee:

    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    # Method to display employee details
    def describe_employee(self):
        print("\n----- Employee Details -----")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)

    # Method to update experience
    def update_year_of_experience(self, new_experience):
        self.year_of_experience = new_experience
        print("Experience updated successfully!")

    # Method to increase salary
    def increase_salary(self, amount):
        self.salary += amount
        print("Salary increased successfully!")


# Creating first object
employee101 = Employee(
    "Ali", 101, "Karachi", 50000,
    "Python Developer", 2022, "BS Computer Science", 2
)

# Printing four attributes individually
print(employee101.name)
print(employee101.number)
print(employee101.address)
print(employee101.salary)

# Calling all methods
employee101.describe_employee()

employee101.increase_salary(10000)
employee101.update_year_of_experience(3)

employee101.describe_employee()


# Creating 10 different employee objects
employee102 = Employee("Ahmed", 102, "Lahore", 45000, "Web Developer", 2021, "BS IT", 3)
employee103 = Employee("Usman", 103, "Karachi", 60000, "Data Analyst", 2020, "BS Data Science", 4)
employee104 = Employee("Hamza", 104, "Islamabad", 55000, "Designer", 2022, "BFA", 2)
employee105 = Employee("Hassan", 105, "Multan", 40000, "Accountant", 2023, "B.Com", 1)
employee106 = Employee("Bilal", 106, "Faisalabad", 70000, "Software Engineer", 2019, "BS CS", 5)
employee107 = Employee("Usama", 107, "Hyderabad", 50000, "HR Manager", 2021, "MBA", 3)
employee108 = Employee("Saad", 108, "Quetta", 65000, "Project Manager", 2018, "MBA", 6)
employee109 = Employee("Zain", 109, "Peshawar", 48000, "Marketing Executive", 2022, "BBA", 2)
employee110 = Employee("Danish", 110, "Sialkot", 75000, "Team Lead", 2017, "MS CS", 7)

# Store all employees in a list
employees = [
    employee101, employee102, employee103, employee104,
    employee105, employee106, employee107, employee108,
    employee109, employee110
]

# Display details of all 10 employees
for employee in employees:
    employee.describe_employee()

# Updating some employees
employee102.increase_salary(5000)
employee103.update_year_of_experience(5)
employee104.increase_salary(8000)
employee105.update_year_of_experience(2)

# Display updated details
employee102.describe_employee()
employee103.describe_employee()
employee104.describe_employee()
employee105.describe_employee()
```
