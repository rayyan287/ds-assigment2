```python
def count_vowels(text):
    count = 0

    for char in text:
        if char in "aeiouAEIOU":
            count += 1

    return count


word = "Welcome"
print(count_vowels(word))
```

```python
def display_student_info(name, age, grade, school):
    print("Name:", name)
    print("Age:", age)
    print("Grade:", grade)
    print("School:", school)


display_student_info(
    name="John Doe",
    age=15,
    grade="10th",
    school="Greenwood High"
)
```

```python
def display_prices(product, **kwargs):
    print("product:", product)

    for website, price in kwargs.items():
        print(website + ": $" + str(price))


display_prices(
    "Laptop",
    Amazon=800,
    eBay=750,
    Walmart=780
)
```

```python
def find_max(*args):
    maximum = args[0]

    for num in args:
        if num > maximum:
            maximum = num

    return maximum


print(find_max(10, 25, 5, 80, 40))
```

```python
delivery_charges = lambda order: 150 if order < 2000 else 0

print(delivery_charges(1500))
print(delivery_charges(2500))
```
